# Import Statements + GPU Setup

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
import tensorflow as tf

# Check if any GPU is detected
print("Num GPUs Available: ", len(tf.config.list_physical_devices('GPU')))

gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        # Set TensorFlow to use only the first GPU
        tf.config.set_visible_devices(gpus[0], 'GPU')
        print(f"Using GPU: {gpus[0]}")
    except RuntimeError as e:
        print(e)
```

```python
model = YOLO("yolov8s.pt")
```

# YOLOv8 Model Tuning

## Retrain Model With More Epochs

```python
model.train(
    data="/home/exh4748/ProjectTortoise/Beta/data.yaml",
    epochs=100,                      # Increased epochs for better convergence
    imgsz=640,                        # Image size
    batch=32,                         # Adjust based on GPU memory
    device=0                          # Use GPU if available
)
```

# YOLO 11 Model Tuning

## Hyperparameter Tuning
- Larger model variant = better accuracy.
- More epochs + Early stopping = balanced convergence.
- Advanced augmentations = better generalization.
- Higher input resolution = improved small object detection.
- Learning rate tuning + AdamW optimizer = smoother convergence.
- Test-time augmentation = higher mAP during inference

```python
model11 = YOLO("yolo11s.pt")  # Small variant for improved capacity

model11.train(
    data="/home/exh4748/ProjectTortoise/Beta/data.yaml",
    epochs=200,
    imgsz=800,              # Higher image resolution
    batch=16,
    lr0=0.0015,
    optimizer="AdamW",
    patience=50,            # More patience for early stopping
    hsv_h=0.015,
    hsv_s=0.6,
    hsv_v=0.4,
    degrees=10.0,
    translate=0.1,
    scale=0.4,
    shear=2.0,
    device=0
)
```

```python
metrics = model11.val()
print(f"\n📊 YOLO11 Tuned Model Metrics")
print(f"mAP@50: {metrics.box.map50:.4f}")
print(f"mAP@50-95: {metrics.box.map:.4f}")
print(f"Precision: {metrics.box.mp:.4f}")
print(f"Recall: {metrics.box.mr:.4f}")
print(f"F1-Score: {metrics.box.f1.mean():.4f}")
```
